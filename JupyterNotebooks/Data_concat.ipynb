```python
def plot_feature(df, cols=None, start=None, end=None, connectgaps=False):
    d = df.copy()

    # ensure datetime index
    d.index = pd.to_datetime(d.index, errors="coerce")
    d = d[~d.index.isna()].sort_index()

    # time filter
    if start is not None or end is not None:
        d = d.loc[start:end]

    # choose columns
    if cols is None:
        cols = d.columns.tolist()

    # numeric convert (important if some columns are strings)
    d[cols] = d[cols].apply(pd.to_numeric, errors="coerce")

    # plot multiple columns
    fig = px.line(d, x=d.index, y=cols,)
    fig.update_traces(connectgaps=connectgaps)
    fig.show()
```

#### Reindexed df

```python
#  build reindexed df
d = df.copy()
d.index = pd.to_datetime(d.index, errors="coerce")
d = d[~d.index.isna()].sort_index()
d = d[~d.index.duplicated(keep="last")]

full_index = pd.date_range(d.index.min(), d.index.max(), freq="5min")
missing_ts = full_index.difference(d.index)
d = d.reindex(full_index)

print("missing timestamps:", len(missing_ts))
print("rows before:", len(df))
print("rows after :", len(d))

# compare missing counts (values)
missing_before = df.isnull().sum().sum()
missing_after  = d.isnull().sum().sum()
print("total missing values before:", missing_before)
print("total missing values after :", missing_after)
```
